```python
import numpy as np
import re

from tensorflow.keras.preprocessing.text import Tokenizer 
from tensorflow import keras
import tensorflow as tf

import matplotlib.pyplot as plt
```

```python
data = """Deep learning (also known as deep structured learning) is part of a broader family of machine learning methods based on artificial neural networks with representation learning. Learning can be supervised, semi-supervised or unsupervised. Deep-learning architectures such as deep neural networks, deep belief networks, deep reinforcement learning, recurrent neural networks, convolutional neural networks and Transformers have been applied to fields including computer vision, speech recognition, natural language processing, machine translation, bioinformatics, drug design, medical image analysis, climate science, material inspection and board game programs, where they have produced results comparable to and in some cases surpassing human expert performance."""
```

```python
sentences = data.split('.')
```

```python
print(sentences)
```

```python
clean_sentences = []
for sentence in sentences:
    if sentence == "":
        continue

    sentence = re.sub("[^A-Za-z0-9]+", " ", sentence)
    sentence = sentence.lower()
    clean_sentences.append(sentence.strip())
```

```python
clean_sentences
```

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts(clean_sentences)

sequences = tokenizer.texts_to_sequences(clean_sentences)

sequences
```

```python
tokenizer.word_index
```

```python
word_to_index = tokenizer.word_index

index_to_word = {}

for word, index in word_to_index.items():
    index_to_word[index] = word

index_to_word
```

```python
vocab_size = len(tokenizer.word_index)+1
context_size = 2
emb_size = 10
```

```python
contexts = []
targets = []

for sequence in sequences:
    for i in range(context_size, len(sequence)-context_size):
        target = sequence[i]
        context = [
            sequence[i-context_size],
            sequence[i-(context_size-1)],
            sequence[i+(context_size-1)],
            sequence[i+context_size]
        ]

        targets.append(target)
        contexts.append(context)
```

```python
len(targets) 
```

```python
## print context-target pairs in words


for i in range(len(targets)): ## print ascii pairs for first 6 targets only
    word_context = []
    for context in contexts[i]: ## single context of a single target at a time
        word_context.append(index_to_word[context])
    print(word_context, '->', index_to_word[targets[i]])
```

```python
X = np.array(contexts)
Y = np.array(targets)
X.shape, Y.shape
```

```python
## now create a model

model = keras.Sequential([
    keras.layers.Input(shape=(4,)),

    keras.layers.Embedding(
        input_dim=len(tokenizer.word_index)+1,
        output_dim=10
    ),

    keras.layers.Lambda(lambda x: tf.reduce_mean(x, axis=1)),

    keras.layers.Dense(512, activation="relu"),
    keras.layers.Dense(256, activation="relu"),
    keras.layers.Dense(vocab_size, activation="softmax")
])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=['accuracy']
)
```

```python
## model training

history = model.fit(
    X,
    Y,
    epochs=80,
    verbose=1
)
```

```python
embeddings = model.get_weights()[0]

print("Embedding shape:", embeddings.shape)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
reduced_embeddings = pca.fit_transform(embeddings)
reduced_embeddings
```

```python
plt.figure(figsize=(14, 10))

for word, index in word_to_index.items():

    x, y = reduced_embeddings[index]

    plt.scatter(x, y)

    plt.annotate(
        word,
        (x, y),
        fontsize=8
    )

plt.title("CBOW Word Embeddings using PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```

```python
test_sentences = [
    "known as structured learning",
    "transformers have applied to",
    "where they produced results",
    "cases surpassing expert performance"
]

for sent in test_sentences:

    words = sent.lower().split()

    missing = [
        word for word in words
        if word not in word_to_index
    ]

    print("Sentence:", sent)
    print("Missing words:", missing)
```

```python
for sent in test_sentences:

    test_words = sent.split()

    x_test = []

    for word in test_words:
        x_test.append(word_to_index.get(word))

    x_test = np.array([x_test])

    pred = model.predict(x_test, verbose=0)

    predicted_index = np.argmax(pred[0])

    print("Context:", test_words)
    print("Predicted word:", index_to_word.get(predicted_index))
    print()
```
